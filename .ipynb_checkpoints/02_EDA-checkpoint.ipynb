# 2. Exploratory Data Analysis — Whole-Leaf Features

This notebook analyzes the features extracted from the **whole leaf mask**.

The main goal is to show how the features contribute to class separation.

## 2.1 Load the Feature Dataset

```python
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.feature_selection import f_classif

DATA_PATH = Path("data") / "citrus_features_8_classes.csv"

df = pd.read_csv(DATA_PATH)

print("Dataset shape:", df.shape)
```

## 2.2 Feature Meaning

### RGB
Red, Green, and Blue are the normal image color channels.

### HSV
- **Hue:** basic color type
- **Saturation:** strength or vividness of the color
- **Value:** brightness

### Grayscale
Grayscale removes color and keeps light/dark intensity.

### Ratios
Green, yellow, and dark ratios describe how much of the whole leaf belongs to those simple color groups.

### Edge density
Edge density describes how much abrupt visual change appears inside the whole leaf.

```python
FEATURE_COLUMNS = [
    "mean_R", "mean_G", "mean_B",
    "std_R", "std_G", "std_B",
    "mean_H", "mean_S", "mean_V",
    "std_H", "std_S", "std_V",
    "green_ratio", "yellow_ratio", "dark_ratio",
    "gray_std", "edge_density"
]

CLASS_ORDER = [
    "Healthy",
    "Fe",
    "Mg",
    "Mn",
    "N",
    "Zn",
    "HLB",
    "Greasy_spot"
]

FEATURE_FAMILIES = {
    "Overall Color": [
        "mean_R", "mean_G", "mean_B",
        "mean_H", "mean_S", "mean_V"
    ],

    "Color Variation": [
        "std_R", "std_G", "std_B",
        "std_H", "std_S", "std_V",
        "gray_std"
    ],

    "Color Ratios": [
        "green_ratio",
        "yellow_ratio",
        "dark_ratio"
    ],

    "Structure": [
        "edge_density"
    ]
}
```

## 2.3 Class Distribution

```python
class_counts = (
    df["class"]
    .value_counts()
    .reindex(CLASS_ORDER)
)

plt.figure(figsize=(7, 4))
plt.bar(
    class_counts.index,
    class_counts.values
)

plt.title("Number of Images per Class")
plt.xlabel("Class")
plt.ylabel("Images")
plt.xticks(rotation=30)

plt.tight_layout()
plt.show()
```

## 2.4 Feature Group Separation

A higher separation score means that a feature group differs more strongly between classes.

```python
X = df[FEATURE_COLUMNS]
y = df["class"]

f_scores, _ = f_classif(
    X,
    y
)

feature_scores = pd.DataFrame({
    "Feature": FEATURE_COLUMNS,
    "Separation Score": f_scores
})

family_rows = []

for family_name, family_features in FEATURE_FAMILIES.items():

    score = feature_scores.loc[
        feature_scores["Feature"].isin(
            family_features
        ),
        "Separation Score"
    ].mean()

    family_rows.append({
        "Feature Group": family_name,
        "Separation Score": round(score)
    })

family_scores = (
    pd.DataFrame(family_rows)
    .sort_values(
        "Separation Score",
        ascending=False
    )
)

plt.figure(figsize=(6, 4))

plt.bar(
    family_scores["Feature Group"],
    family_scores["Separation Score"]
)

plt.title("Class Separation by Feature Group")
plt.ylabel("Separation Score")
plt.xticks(rotation=25)

plt.tight_layout()
plt.show()

print(
    family_scores.to_string(index=False)
)
```

## 2.5 Green and Yellow Area by Class

```python
ratio_means = (
    df.groupby("class")[
        ["green_ratio", "yellow_ratio"]
    ]
    .mean()
    .reindex(CLASS_ORDER)
    * 100
)

x = np.arange(len(CLASS_ORDER))
width = 0.38

plt.figure(figsize=(8, 4))

plt.bar(
    x - width / 2,
    ratio_means["green_ratio"],
    width,
    label="Green %"
)

plt.bar(
    x + width / 2,
    ratio_means["yellow_ratio"],
    width,
    label="Yellow %"
)

plt.xticks(
    x,
    CLASS_ORDER,
    rotation=30
)

plt.ylabel("Whole Leaf Area (%)")
plt.title("Average Green and Yellow Area by Class")
plt.legend()

plt.tight_layout()
plt.show()
```

## 2.6 Brightness Variation by Class

`gray_std` measures how uneven the light/dark appearance is across the whole leaf.

```python
gray_means = (
    df.groupby("class")["gray_std"]
    .mean()
    .reindex(CLASS_ORDER)
)

plt.figure(figsize=(7, 4))

plt.bar(
    gray_means.index,
    gray_means.values
)

plt.title("Average Brightness Variation by Class")
plt.ylabel("Gray Standard Deviation")
plt.xticks(rotation=30)

plt.tight_layout()
plt.show()
```

## 2.7 Average Hue by Class

Hue provides a simple summary of the overall color tendency.

```python
hue_means = (
    df.groupby("class")["mean_H"]
    .mean()
    .reindex(CLASS_ORDER)
)

plt.figure(figsize=(7, 4))

plt.bar(
    hue_means.index,
    hue_means.values
)

plt.title("Average Hue by Class")
plt.ylabel("Mean Hue")
plt.xticks(rotation=30)

plt.tight_layout()
plt.show()
```

## 2.8 Edge Density by Class

```python
edge_means = (
    df.groupby("class")["edge_density"]
    .mean()
    .reindex(CLASS_ORDER)
    * 100
)

plt.figure(figsize=(7, 4))

plt.bar(
    edge_means.index,
    edge_means.values
)

plt.title("Average Edge Percentage by Class")
plt.ylabel("Edge Pixels (%)")
plt.xticks(rotation=30)

plt.tight_layout()
plt.show()
```

## 2.9 Two Features Together

The scatter plot helps show class overlap.

Each point represents one complete leaf.

```python
plt.figure(figsize=(7, 5))

for class_name in CLASS_ORDER:

    subset = df[
        df["class"] == class_name
    ]

    plt.scatter(
        subset["yellow_ratio"] * 100,
        subset["gray_std"],
        label=class_name,
        alpha=0.55,
        s=22
    )

plt.xlabel("Yellow Area (%)")
plt.ylabel("Brightness Variation")
plt.title("Yellow Area vs Brightness Variation")
plt.legend(fontsize=8, ncol=2)

plt.tight_layout()
plt.show()
```

## 2.10 Top Individual Features

```python
top_features = (
    feature_scores
    .sort_values(
        "Separation Score",
        ascending=False
    )
    .head(6)
    .copy()
)

top_features["Separation Score"] = (
    top_features["Separation Score"]
    .round()
    .astype(int)
)

plot_data = top_features.sort_values(
    "Separation Score"
)

plt.figure(figsize=(6, 4))

plt.barh(
    plot_data["Feature"],
    plot_data["Separation Score"]
)

plt.title("Top Features for Class Separation")
plt.xlabel("Separation Score")

plt.tight_layout()
plt.show()

print(top_features.to_string(index=False))
```

# EDA Conclusion

The whole-leaf measurements can be interpreted using these feature groups:

- overall color gives a general color clue;
- color ratios show how much of the leaf is green or yellow;
- variation features show how uneven the leaf is;
- edge density adds structural information.

These charts describe how the selected whole-leaf color and structure features vary across the eight classes.
